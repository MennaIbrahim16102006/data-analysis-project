```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/dirty_cafe_sales.csv")
```

```python
df.head()
```

```python
df.info()
```

معرفه القيم المفقوده  (cleaning اول خطوه فى ال )

```python
df.isnull().sum()
```

 بقيم مفقوده unknown  /error  استبدال كلمه

```python
df.replace(["ERROR","UNKNOWN"],
np.nan, inplace=True)
```

```python
df.isnull().sum()
```

حذف الصفوف المتكرره

```python
df.drop_duplicates(inplace=True)
```

تحويل الاعمده الرقميه

```python
df["Quantity"]=pd.to_numeric(df["Quantity"],errors="coerce")
df["Price Per Unit"]=pd.to_numeric(df["Price Per Unit"],errors="coerce")
df["Total Spent"]=pd.to_numeric(df["Total Spent"],errors="coerce")
```

تحويل التاريخ

```python
df["Transaction Date"]=pd.to_datetime(df["Transaction Date"],errors="coerce")
```

median ملىء القيم الناقصه فى الاعمده الرقميه بال

```python
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Price Per Unit"] = df["Price Per Unit"].fillna(df["Price Per Unit"].median())
df["Total Spent"] = df["Total Spent"].fillna(df["Total Spent"].median())
```

modeملىء القيم الناقصه فى الاعمده النصيه باسخدام ال

```python
df["Item"] = df["Item"].fillna(df["Item"].mode()[0])
df["Payment Method"] = df["Payment Method"].fillna(df["Payment Method"].mode()[0])
df["Location"] = df["Location"].fillna(df["Location"].mode()[0])
df["Transaction Date"] = df["Transaction Date"].fillna(df["Transaction Date"].mode()[0])
```

حذف اى صف مازال تاريخه مفقود

```python
df.dropna(subset=["Transaction Date"], inplace=True)
```

التاكد من النتيجه النهائيه

```python
df.info()
df.isnull().sum()
```

حفظ الملف بعد التنظيف

```python
df.to_csv("cleaned_cafe_sales.csv", index=False)
```
